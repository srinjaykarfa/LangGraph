```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
    runs:int
    balls:int
    fours:int
    sixes:int
    strike_rate:float
    boundary_rate:float
    balls_per_boundary:float
    summary:str
```

```python
def strikeRate(state:BatsmanState)->BatsmanState:
    strikeRate = (state['runs'] / state['balls']) * 100
    return {'strike_rate':strikeRate}
```

```python
def boundaryRate(state:BatsmanState)->BatsmanState:
    boundaryRate = (state['fours']*4 + state['sixes']*6) / state['runs'] * 100
    return {'boundary_rate':boundaryRate}
```

```python
def ballsPerBoundary(state:BatsmanState)->BatsmanState:
    ballsPerBoundary = state['balls']/(state['fours'] + state['sixes'])
    return {'balls_per_boundary':ballsPerBoundary}
```

```python
def summary(state:BatsmanState)->BatsmanState:
    summary= f"""Runs: {state['runs']},\n
    Balls: {state['balls']},\n 
    Fours: {state['fours']},\n 
    Sixes: {state['sixes']},\n
    Strike Rate: {state['strike_rate']:.2f},\n
    Boundary Rate: {state['boundary_rate']:.2f}%, \n
    Balls per Boundary: {state['balls_per_boundary']:.2f}"""
    return {'summary':summary}
```

```python
graph=StateGraph(BatsmanState)

# nodes
graph.add_node('calculate_sr',strikeRate)
graph.add_node('calculate_br',boundaryRate)
graph.add_node('calculate_bpb',ballsPerBoundary)
graph.add_node('summary',summary)

# edges
graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_br')
graph.add_edge(START,'calculate_bpb')
graph.add_edge('calculate_sr','summary')
graph.add_edge('calculate_br','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('summary',END)

# compile
workflow=graph.compile()
```

```python
# execute
initial_state={
    'runs': 220,
    'balls': 80,
    'fours': 18,
    'sixes': 15
}
final_state=workflow.invoke(initial_state)
print(final_state['summary'])
```
