```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal
```

```python
class QuadState(TypedDict):
    a: int
    b: int
    c: int
    equation: str
    discriminant: float
    result: str
```

```python
def show_equation(state: QuadState) -> QuadState:
    equation = f"{state['a']}x^2 + {state['b']}x + {state['c']} = 0"
    return {'equation': equation}
```

```python
def calculate_discriminant(state: QuadState) -> QuadState:
    discriminant = state['b']**2 - 4*state['a']*state['c']
    return {'discriminant': discriminant}
```

```python
def real_roots(state:QuadState) -> QuadState:
    root1 = (-state['b'] + state['discriminant']**0.5) / (2*state['a'])
    root2 = (-state['b'] - state['discriminant']**0.5) / (2*state['a'])
    return {'result': f"Real roots: {root1}, {root2}"}
```

```python
def repeated_roots(state:QuadState) -> QuadState:
    root = -state['b'] / (2*state['a'])
    return {'result': f"Repeated root: {root}"}
```

```python
def complex_roots(state:QuadState) -> QuadState:
    real_part = -state['b'] / (2*state['a'])
    imaginary_part = (abs(state['discriminant'])**0.5) / (2*state['a'])
    return {'result': f"Complex roots: {real_part} ± {imaginary_part}i"}
```

```python
def check_discriminant(state: QuadState) -> Literal['real_roots','repeated_roots','complex_roots']:
    if state['discriminant'] > 0:
        return 'real_roots'
    elif state['discriminant'] == 0:
        return 'repeated_roots'
    else:
        return 'complex_roots'
```

```python
graph = StateGraph(QuadState)

# add nodes
graph.add_node("show_equation",show_equation)
graph.add_node("calculate_discriminant",calculate_discriminant)
graph.add_node('real_roots',real_roots)
graph.add_node('repeated_roots',repeated_roots)
graph.add_node('complex_roots',complex_roots)


# add edges
graph.add_edge(START,"show_equation")
graph.add_edge("show_equation","calculate_discriminant")
graph.add_conditional_edges("calculate_discriminant",check_discriminant)
graph.add_edge("real_roots",END)
graph.add_edge("repeated_roots",END)
graph.add_edge("complex_roots",END)

# compile graph
workflow=graph.compile()
```

```python
# execute graph
initial_state={
    'a': 4,
    'b': 4,
    'c': 1
}
final_state = workflow.invoke(initial_state)
print(final_state)
```
